## Building the Tokenizer

**Tokenizer Interface**



**Constructor**:- Receive a vocabulary a list of characters that the tokenizer can process

**Methods**

train_from_text: Creates a tokenizer from the text.

encode: Converts a string into a list of token IDs.

decode: Converts a list of token IDs back into text.

vocabulary_size: Returns the size of the vocabulary.



```python
class CharTokenizer:
  """
  This class will create a tokenizer for provided text, encode and decode the text and claculate the total length of vacabulary
  """
  def __init__(self, vocabulary:str) -> None:
    self.token_id_for_char = {char : token_id for token_id, char in enumerate(vocabulary)}
    self.char_for_token_id = {token_id : char for token_id, char in enumerate(vocabulary)}

  @staticmethod
  def train_from_text(text:str):
    # remove the duplicay from text
    vocabulary = set(text)
    return CharTokenizer(sorted(vocabulary))

  def encode(self, text:str):
    token_ids = []
    for char in text:
      token_ids.append(self.token_id_for_char[char])

    return  torch.tensor(token_ids, dtype=torch.long)

  def decode(self,token_ids):
    char_list = []
    for token_id in token_ids.tolist():
      char = self.char_for_token_id[token_id]
      char_list.append(char)
    return "".join(char_list)

  def vocabulary_size(self):
    return len(self.token_id_for_char)
```

```python
# Testing the code
tokenizer = CharTokenizer.train_from_text("Hello Prashant")
print(tokenizer)
print(tokenizer.encode("Hello Prashant"))
print(tokenizer.decode(tokenizer.encode("Hello Prashant")))
```
